# Объект Series

Series - одномерный, похожий на массив объект, который содержит последовательность данных и ассоциированный с ним
массив меток, который называется индексом.

Самый просто объект Series состоит только из массива данных:

```python
import pandas as pd
from pandas import Series, DataFrame

obj = Series([4, 7, -5, 3])
obj
```

В строковом представлении Series, отображаемом в интерактивном режиме, индекс находится слева, a значения - справа.
Так как не был задан индекс для данных, по умолчанию создается индекс, который состоит из целых чисел от 0 до N. N -
это длина массива данных.

Получение представление самого массива и его индекса можно с помощью атрибутов values и index:

```python
obj.values
```

```python
obj.index # как range(4)
```

Создание объекта Series с индексом, который идентифицирует каждый элемент данных:

```python
import pandas as pd
from pandas import Series, DataFrame

obj2 = Series([4, 7, -5, 3], index=['d', 'b', 'a', 'c'])
obj2
```

```python
obj2.index
```

В отличие от массивов NumPy, для выборки одного или нескольких значений можно использовать метки в индексе:

```python
obj2['a']
```

```python
obj2['d'] = 6
```

```python
obj2[['c', 'a', 'd']]
```

В данном случае ['c', 'a', 'd'] интерпретируется как список индексов.

Функции NumPy, такие, как фильтрация с помощью булева массива, скалярное умножение или применение
математических функций, сохраняют связь между индексом и значением:

```python
obj2[obj2 > 0]
```

```python
obj2 * 2
```

```python
import numpy as np

np.exp(obj2)
```

Объект Series можно представить как упорядоченный словарь фиксированной длины, так как он отображает индекс на
данные. Его можно передавать функциям, ожидающим получить словарь:

```python
'b' in obj2
```

```python
'e' in obj2
```

Если имеется словарь Python, содержащий данные, то из него можно создать объект Series:

```python
import pandas as pd
from pandas import Series, DataFrame

sdata = {'Ohio': 35000, 'Texas': 71000, 'Oregon': 16000, 'Utah': 5000}
obj3 = pd.Series(sdata)
obj3
```

Если передается только словарь, то в получившемся объекте Series ключи будут храниться в индексе по порядку:

```python
states = ['California', 'Ohio', 'Oregon', 'Texas']
obj4 = pd.Series(sdata, index=states)
obj4
```

В данном случае три значения, найденных в sdata, помещены в соответствующие им позиции, a для метки 'California'
никакого значения не нашлось, поэтому ей соответствует признак NaN (не число), которым в pandas обозначаются
отсутствующие значения. Поскольку строки 'Utah' не было в списке states, то ее нет и в результирующем объекте.

Для распознавания отсутствующих данных в pandas следует использовать функции isnull и notnull:

```python
pd.isnull(obj4)
```

```python
pd.notnull(obj4)
```

У объекта Series есть методы экземпляра:

```python
obj4.isnull()
```

При выполнении арифметических операций объект Series автоматически выравнивает данные по индексной метке:

```python
obj3
```

```python
obj4
```

```python
obj3 + obj4
```

У объекта Series, и у его индекса имеется атрибут name, тесно связанный с другими частями функциональности pandas:

```python
obj4.name = 'population'
```

```python
obj4.index.name = 'state'
```

```python
obj4
```

Индекс объекта Series можно изменить на месте с помощью присваивания:

```python
obj
```

```python
obj.index = ['Bob', 'Steve', 'Jeff', 'Ryan']
obj
```

## Объект DataFrame

Объект DataFrame представляет табличную структуру данных, состоящую из упорядоченной коллекции столбцов, причем
типы значений (числовой, строковый, булев и т.д.) в разных столбцах могут различаться. В объекте DataFrame хранятся
два индекса: по строкам и по столбцам. Можно считать, что это словарь объектов Series, имеющих общий индекс. Внутри
объекта данные хранятся в виде одного или нескольких двумерных блоков, а не в виде списка, словаря или еще какой-нибудь коллекции одномерных массивов.

Есть много способов сконструировать объект DataFrame, один из самых распространенных – на основе словаря списков
одинаковой длины или массивов NumPy:

```python
import pandas as pd
from pandas import Series, DataFrame

data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],
       'year': [2000, 2001, 2002, 2001, 2002, 2003],
       'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
frame = pd.DataFrame(data)
```

Получившемуся DataFrame автоматически будет назначен индекс, как и в случае Series, и столбцы расположатся по
порядку:

```python
frame
```

В Jupyter-блокноте объекты DataFrame отображаются в виде HTML-таблицы, более удобной для браузера. Для больших
объектов DataFrame метод head отбирает только первые пять строк:

```python
frame.head()
```

Если задать последовательность столбцов, то столбцы DataFrame расположатся строго в указанном порядке:

```python
pd.DataFrame(data, columns=['year', 'state', 'pop'])
```

Если запросить столбец, которого нет в data, то он будет заполнен значениями NaN:

```python
frame2 = pd.DataFrame(data, columns=['year', 'state', 'pop', 'debt'],
                      index=['one', 'two', 'three', 'four',
                            'five', 'six'])
```

```python
frame2
```

```python
frame2.columns
```

Столбец DataFrame можно извлечь как объект Series, воспользовавшись нотацией словарей или помощью атрибута:

Синтаксис frame2[column] работает для любого имени столбца, a frame2.column – только когда имя столбца – допустимое
имя переменной Python.

```python
frame2['state']
```

```python
frame2.year
```

Возвращенный объект Series имеет тот же индекс, что и DataFrame, a его атрибут name установлен соответствующим
образом.

Строки также можно извлечь по позиции или по имени с помощью специального атрибута loc (подробнее об этом ниже):

```python
frame2.loc['three']
```

Столбцы можно модифицировать путем присваивания. Например, пустому столбцу 'debt' можно присвоить скалярное
значение или массив значений:

```python
frame2['debt'] = 16.5
frame2
```

```python
frame2['debt'] = np.arange(6.)
frame2
```

Когда столбцу присваивается список или массив, длина значения должна совпадать с длиной DataFrame. Если же
присваивается объект Series, то его метки будут точно выровнены с индексом DataFrame, a в «дырки» будут
вставлены значения NaN:

```python
val = pd.Series([-1.2, -1.5, -1.7], index=['two', 'four', 'five'])
frame2['debt'] = val
frame2
```

Присваивание несуществующему столбцу приводит к созданию нового столбца. Для удаления столбцов служит ключевое
слово del, как и в обычном словаре.

Для демонстрации работы del добавим новый столбец булевых признаков, показывающих, находится ли в столбце state
значение 'Ohio':

```python
frame2['eastern'] = frame2.state == 'Ohio'
frame2
```

Новый столбец нельзя создать, пользуясь синтаксисом frame2.eastern.

Для удаления этого столбца воспользуемся методом del:

```python
del frame2['eastern']
frame2.columns
```

```python
frame2.columns
```

Столбец, возвращенный в ответ на запрос к DataFrame по индексу, является представлением, a не копией данных.
Следовательно, любые модификации этого объекта Series найдут отражение в DataFrame. Чтобы скопировать столбец,
нужно вызвать метод copy объекта Series.

Еще одна распространенная форма данных – словарь словарей:

```python
pop = {'Nevada': {2001: 2.4, 2002: 2.9},
      'Ohio': {2000: 1.5, 2001: 1.7, 2002: 3.6}}
```

Если передать вложенный словарь объекту DataFrame, то pandas интерпретирует ключи внешнего словаря как столбцы, a
ключи внутреннего словаря – как индексы строк:

```python
frame3 = DataFrame(pop)
frame3
```

Объект DataFrame можно транспонировать (переставить строки и столбцы), воспользовавшись таким же синтаксисом, как для словаря
NumPy:

```python
frame3.T
```

Ключи внутренних словарей объединяются и сортируются для образования индекса результата. Однако этого не
происходит, если индекс задан явно:

```python
DataFrame(pop, index=[2001, 2002, 2003])
```

Словари объектов Series интерпретируются очень похоже:

```python
pdata = {'Ohio': frame3['Ohio'][:-1],
        'Nevada':frame3['Nevada'][:2]}
DataFrame(pdata)
```

Если у объектов, возвращаемых при обращении к атрибутам index и columns объекта DataFrame, установлен атрибут
name, то он также выводится:

```python
frame3.index.name = 'year'; frame3.columns.name = 'state'
frame3
```

Как и в случае Series, атрибут values возвращает данные, хранящиеся в DataFrame, в виде двумерного массива ndarray:

```python
frame3.values
```

Если у столбцов DataFrame разные типы данных, то dtype массива values будет выбран так, чтобы охватить все столбцы:

```python
frame2.values
```

## Индексные объекты

В индексных объектах pandas хранятся метки вдоль осей и прочие метаданные (например, имена осей). Любой массив
или иная последовательность меток, указанная при конструировании Series или DataFrame, преобразуется в объект Index:

```python
import pandas as pd
from pandas import Series

obj = Series(range(3), index=['a', 'b', 'c'])
index = obj.index
index
```

```python
index[1:]
```

Индексные объекты неизменяемы, т.е. пользователь не может их модифицировать:

```python
index[1] = 'd'
```

Неизменяемость важна, для того чтобы несколько структур данных могли совместно использовать один и тот же
индексный объект, не опасаясь его повредить:

```python
labels = pd.Index(np.arange(3))
labels
```

```python
obj2 = pd.Series([1.5, -2.5, 0], index=labels)
obj2
```

```python
obj2.index is labels
```

Индексный объект не только похож на массив, но и ведет себя как множество фиксированного размера:

```python
frame3
```

```python
frame3.columns
```

```python
'Ohio' in frame3.columns
```

```python
2003 in frame3.index
```

В отличие от множеств Python, индекс в pandas может содержать повторяющиеся метки:

```python
dup_labels = pd.Index(['foo', 'foo', 'bar', 'bar'])
dup_labels
```

Если при выборе указана повторяющаяся метка, то будут выбраны все элементы с такой меткой.

У любого объекта Index есть ряд свойств и методов для ответа на типичные вопросы о хранящихся в нем данных.

Методы и свойства объекта Index: append; diff; intersection; union; isin; delete; drop; insert; is_monotonic; is_unique; unique.

## Задание 2

Приведите практические примеры всех методов и свойств объекта Index со своими примерами, перечисленных выше. В
отдельной строке перед примером необходимо дать описание каждого метода и свойства.

```python
# append - добавление нового элемента
labels = pd.Index(["label 1", "label 2"])
new_labels = labels.append(pd.Index(["label 3"]))
new_labels
```

```python
# diff - вычисление разности между элементами (параметр настраиваемый)
indices = pd.Index([0, 2, 4, 6, 8, 9, 10])
indices.diff()
```

```python
# intersection - пересечение значений двух объектов Index
indices2 = pd.Index([2, 4, 6, 8])
indices.intersection(indices2)
```

```python
# union - объединение значений двух объектов Index
indices.union(indices2)
```

```python
# isin - проверка вхождения элемента в Index
indices.isin([2, 4, 6, 8])
```

```python
# delete - удаление элемента по позиции
indices.delete(1)
```

```python
# drop - удаление элемента по значению
indices.drop(2)
```

```python
# insert - Добавление элемента на позицию
indices.insert(1, 1)
```

```python
# is_monotonic - проверка монотонности Index
print(indices.is_monotonic_increasing)
print(indices.is_monotonic_decreasing)
```

```python
# is_unique - проверка уникальности элементов внутри Index
indices.is_unique
```

```python
# unique - возвращает уникальные элементы Index
indices.unique()
```
