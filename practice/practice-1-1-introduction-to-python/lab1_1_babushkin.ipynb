# Операторы Python

| Операция | Результат                               |
|----------|-----------------------------------------|
| `x + y`  | Сложение (сумма x и y)                  |
| `x - y`  | Вычитание (разница между x и y)         |
| `-x`     | Смена знака x                           |
| `+x`     | Тождественность x                       |
| `x * y`  | Умножение x на y                        |
| `x / y`  | Деление x на y                          |
| `x // y` | Получение целой части от деления x на y |
| `x % y`  | Остаток от деления x / y                |
| `x ** y` | Возведение в степень                    |
